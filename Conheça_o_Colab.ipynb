```
from pypdf import PdfReader
from google.colab import files

def converter_meu_pdf():
    try:
        # Abre o PDF gerado
        leitor = PdfReader("documento_teste.pdf")

        # Cria o arquivo txt
        with open("resultado_final.txt", "w", encoding="utf-8") as arquivo_txt:
            for i, pagina in enumerate(leitor.pages):
                texto = pagina.extract_text()
                if texto:
                    arquivo_txt.write(texto)

        print("Sucesso! O texto foi extraído.")
        print("Baixando seu arquivo 'resultado_final.txt'...")

        # Baixa o txt pro seu computador
        files.download("resultado_final.txt")

    except FileNotFoundError:
        print("Erro: Certifique-se de que o arquivo 'documento_teste.pdf' foi arrastado para a barra lateral do Colab!")
    except Exception as e:
        print(f"Ocorreu um erro: {e}")

# Executa a função
converter_meu_pdf()
```

```
from pypdf import PdfReader
from google.colab import files

def converter_pdf_dinamico():
    # 1. Pede para você digitar o nome do arquivo na tela
    nome_do_arquivo = input("Digite o nome exato do arquivo PDF (ex: documento.pdf): ")

    # Garante que o usuário digitou o '.pdf' no final, para evitar erros bobos
    if not nome_do_arquivo.lower().endswith('.pdf'):
        nome_do_arquivo += '.pdf'

    try:
        print(f"\nTentando ler o arquivo: {nome_do_arquivo}...")
        leitor = PdfReader(nome_do_arquivo)

        # Define o nome do arquivo de texto baseado no nome do PDF
        nome_txt_saida = nome_do_arquivo.replace('.pdf', '.txt')

        with open(nome_txt_saida, "w", encoding="utf-8") as arquivo_txt:
            for i, pagina in enumerate(leitor.pages):
                texto = pagina.extract_text()
                if texto:
                    arquivo_txt.write(texto)

        print(f"Sucesso! Texto extraído com sucesso.")
        print(f"Baixando seu arquivo '{nome_txt_saida}'...")

        # Baixa o arquivo txt gerado
        files.download(nome_txt_saida)

    except FileNotFoundError:
        print(f"\n❌ ERRO: O arquivo '{nome_do_arquivo}' não foi encontrado.")
        print("Verifique se você arrastou o PDF para a barra lateral esquerda do Colab e se digitou o nome correto.")
    except Exception as e:
        print(f"\n❌ Ocorreu um erro inesperado: {e}")

# Executa o programa
converter_pdf_dinamico()
```

```
from pdf2docx import Converter
from google.colab import files
import os

def converter_pdf_para_word_ajustado():
    # Pede o nome do arquivo
    nome_do_arquivo = input("Digite o nome do arquivo PDF: ")

    # Garante que termine com .pdf e remove espaços extras que você possa digitar sem querer
    nome_do_arquivo = nome_do_arquivo.strip()
    if not nome_do_arquivo.lower().endswith('.pdf'):
        nome_do_arquivo += '.pdf'

    nome_word_saida = nome_do_arquivo.replace('.pdf', '.docx')

    # Verifica se o arquivo existe usando o 'os' correto em minúsculo
    if not os.path.exists(nome_do_arquivo):
        print(f"\n❌ ERRO: O arquivo '{nome_do_arquivo}' não foi encontrado na barra lateral.")
        print("Arquivos disponíveis que eu achei aí no seu menu: novo_teste.pdf, documento_teste.pdf")
        return

    try:
        print(f"\nIniciando conversão de '{nome_do_arquivo}'...")
        cv = Converter(nome_do_arquivo)
        cv.convert(nome_word_saida, start=0, end=None)
        cv.close()

        print(f"\n✨ Pronto! Baixando seu arquivo Word: {nome_word_saida}")
        files.download(nome_word_saida)

    except Exception as e:
        print(f"\n❌ Ocorreu um erro: {e}")

# Executa
converter_pdf_para_word_ajustado()
```
